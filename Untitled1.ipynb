### 📊 Calculate Quartiles & IQR

- 🔹 **Q1 (25th Percentile):** `np.percentile(AudienceScore, 25)` finds the value below which 25% of the scores fall.
- 🔹 **Q3 (75th Percentile):** `np.percentile(AudienceScore, 75)` finds the value below which 75% of the scores fall.
- 📏 **IQR:** `Q3 - Q1` calculates the **Interquartile Range**, which represents the spread of the middle 50% of audience scores.
- 🖨️ **Print:** Displays Q1, Q3, and IQR values.


```python
Q1 = np.percentile(AudienceScore, 25)
Q3 = np.percentile(AudienceScore, 75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

### 🚨 Detect Outliers Using IQR Method

- 🧹 **Remove Missing Values:** `~np.isnan(AudienceScore)` removes `NaN` values before calculating quartiles.
- 🔹 **Q1:** Calculates the 25th percentile of the valid audience scores.
- 🔹 **Q3:** Calculates the 75th percentile of the valid audience scores.
- 📏 **IQR:** `Q3 - Q1` calculates the Interquartile Range.
- ⬇️ **Lower Limit:** `Q1 - 1.5 × IQR` defines the lower boundary for detecting outliers.
- ⬆️ **Upper Limit:** `Q3 + 1.5 × IQR` defines the upper boundary for detecting outliers.
- 🚨 **Outliers:** Scores below the lower limit or above the upper limit are identified as outliers.
- 🔢 **Number of Outliers:** `len(outliers)` counts the total number of outlier scores.



```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

scores = AudienceScore[~np.isnan(AudienceScore)]
Q1 = np.percentile(scores, 25)
Q3 = np.percentile(scores, 75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = scores[(scores < lower_limit) | (scores > upper_limit)]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:", outliers)
print("Number of Outliers:", len(outliers))

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:", outliers)
print("Number of Outliers:", len(outliers))
```
